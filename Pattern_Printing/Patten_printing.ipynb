## FOUR 4 steps for pattern printing
- 1. outer loop is for number of lines ( rows)
- 2. inner loop is for number of columns (and connect them to rows )
- 3. always print * inside the inner loops 
- 4. symmetry observation for complex patterns

```python
# example 

for i in range(5):
    for j in range(5):
        print('*', end=' ')
    print('')
```

```python
for  i in range(10):
    for j in range(10):
        print('*', end=' ')
    print()
```
