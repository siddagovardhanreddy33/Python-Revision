### 3.Create a function to compute the factorial of a number recursively

```python
def factorial(x):
    if x ==0 or x==1:
        return 1
    else:
        return x * factorial(x-1)
```

### 9. Implement a basic calculator using functions. 


```python
def calculator(x,y,operator):
    if operator == '+':
        return x+y
    
    elif operator == '-':
        return x-y
    
    elif operator == '*':
        return x*y
    
    elif operator == '/':
        return x/y
    
    elif operator == '//':
        return x//y
    
    elif operator == '%':
        return x%y
    
    elif operator == '**' or '^':
        return x ** y
    else :
        return 'wrong operator'
```

```python
calculator(24,4,'*')
```

```python
def mergedicts(d1, d2):
    result = d1.copy()

    for key, value in d2.items():
        if key in result:
            result[key] = [result[key], value]
        else:
            result[key] = value

    return result
```

```python

a = {
    'name': 'tillu',
    'age': '22',
    'field': 'IT'
}

b = {
    'name': 'vinnu',
    'age': 22,
    'field': 'real estate'
}

print(mergedicts(a, b))
```

### 11. Count the frequency of each word in a sentence.

### 14. Write a function to check if a number is a prime.

### 21. Reduce a list to its sum using reduce(). 


```python
from functools import reduce

# List of numbers
numbers = [1, 2, 3, 4, 5]

# Find the sum using reduce()
total = reduce(lambda x, y: x + y, numbers)

print(total)
```

```python
from functools import reduce

def sum_list(lst):
    return reduce(lambda x, y: x + y, lst)
```

```python
numbers = [10, 20, 30, 40]
print(sum_list(numbers))
```
