```python
n = 5

for i in range(1, n + 1):
    for j in range(1, i + 1):

        if j % 2 != 0:      # odd position
            print(j, end="")
        else:               # even position
            if j % 4 == 0:
                print("*", end="")
            else:
                print("#", end="")

    print()
```

```python
pattern = ['1', '#', '3', '*', '5']

for i in range(1, len(pattern) + 1):
    for j in range(i):
        print(pattern[j], end="")
    print()
```

```python
text = input("Enter a string: ")
print(text)

# Reverse the string
reverse_text = text[::-1]

if text == reverse_text:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
s = input()
print(s)

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
num = int(input("Enter a number: "))
print(num)

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
n = 5

for i in range(1, n + 1):

    # Print spaces
    for j in range(n - i):
        print(" ", end="")

    # Print stars
    for k in range(2 * i - 1):
        print("*", end="")

    print()
```
